```python
df_ml = df.copy()
```

```python
sleep_dictionary = {"Poor": 0, "Fair": 1, "Good": 2}

social_dictionary = {"low": 0, "medium": 1, "high": 2}

df_ml["sleep_quality"] = df_ml["sleep_quality"].map(sleep_dictionary)
df_ml["social_interaction_level"] = df_ml["social_interaction_level"].map(social_dictionary)

print(df_ml[["sleep_quality", "social_interaction_level"]].head())
```

```python
df_ml
```

```python
df_ml.info()
```

```python
# ==============================================================================
# FEATURE ENGINEERING: Handling Data Leakage
# ==============================================================================
# The 'mental_health_risk_score' is highly correlated with our potential target 
# variables (like depression or anxiety) because it was likely calculated FROM them.
# Keeping it in our training set would cause 'Data Leakage', allowing the model 
# to cheat and leading to falsely high accuracy scores.

# Dropping the leakage column from our Machine Learning dataframe
df_ml = df_ml.drop(columns=['mental_health_risk_score'])

# Verifying the drop operation by checking the remaining columns
print("Remaining columns in df_ml:")
print(df_ml.columns.tolist())
```

```python
# ==============================================================================
# FEATURE ENGINEERING: Final Missing Value Check
# ==============================================================================
# Machine learning models (like SVM, KNN, Random Forest) will throw an error 
# if they encounter any NaN values. We must ensure a 100% clean matrix.

# Checking the total sum of missing values for each column
print("Missing values in df_ml:")
print(df_ml.isnull().sum())
```

```python
# ==============================================================================
# EXPORTING THE CLEAN DATASET
# ==============================================================================
# Saving our fully preprocessed and encoded dataset for future Machine Learning tasks.

df_ml.to_csv("cleaned_mental_health_data.csv", index=False)
print("Clean dataset successfully saved as 'cleaned_mental_health_data.csv'!")
```
